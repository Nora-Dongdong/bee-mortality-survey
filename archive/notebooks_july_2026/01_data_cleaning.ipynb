## March data

```python
import pandas as pd

# 1. LOAD THE DATA
# We skip row 2 (index 1) because it contains the text definitions, not data.
print("Loading raw data...")
df = pd.read_csv('../data/raw/data/2026_surveydata_mar9_v1.csv', skiprows=[1])
print(f"Original shape: {df.shape}")

# 2. REMOVE SUMMARY ROWS (The junk at the bottom)
# The summary rows at the bottom don't have an 'id_location'. 
# By dropping rows where 'id_location' is empty (NaN), we instantly delete the bottom junk.
df = df.dropna(subset=['id_location'])
print(f"Shape after removing summary rows: {df.shape} ")

# 3. CONVERT COLUMNS TO NUMERIC
# Sometimes pandas reads numbers as text if there were weird characters. 
# This forces our key columns to be strictly numeric.
key_columns = ['2025_winter_in', '2026_winter_uit', 'imker_methode', 'winter_food', 'varroa_treatment']
for col in key_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# 4. CREATE DUMMY VARIABLES FOR REGRESSION
# Regression models cannot understand "Category 3". They only understand "Is Category 3? Yes (1) or No (0)".
# This creates new columns with 1s and 0s for your categories.
print("Generating dummy variables for regression...")

# Create dummies for Beekeeping Method
methode_dummies = pd.get_dummies(df['imker_methode'], prefix='methode', dummy_na=False)*1
# Create dummies for Winter Food
food_dummies = pd.get_dummies(df['winter_food'], prefix='food', dummy_na=False)*1
# Create dummies for Varroa Treatment
varroa_dummies = pd.get_dummies(df['varroa_treatment'], prefix='varroa', dummy_na=False)*1

# Join the new dummy columns back to the main dataset
df_clean = pd.concat([df, methode_dummies, food_dummies, varroa_dummies], axis=1)

# 5. SAVE THE CLEANED DATA
# We save this as a NEW file. Your notebooks will use this clean file from now on.
output_path = 'data/2026_surveydata_cleaned_v1.csv'
df_clean.to_csv(output_path, index=False)
print(f"\nSuccess! Clean data saved to: {output_path}")
print(f"Final dataset has {df_clean.shape[0]} rows and {df_clean.shape[1]} columns.")
```
